```python
# Inicialmente, para poder importar y graficar más fácilmente los datos del conjunto de datos (DataSet) es necesario
# importar varias librerías, así:
import pandas as pd
# La librería Pandas se usa para la manipulación y análisis de conjuntos de datos.
# Pueden consultar la información oficial de la librería en "https://pandas.pydata.org/"
import matplotlib.pyplot as plt
# La librería matplotlib, se usa para la generación de gráficos en 2D
# https://matplotlib.org/
```

```python
#Cargar el conjunto de datos COVID-19 desde el repositorio externo.
#Usando Pandas, se importan los datos en una estructura tipo DataFrame para una mejor manipulación

url = 'https://raw.githubusercontent.com/CSSEGISandData/COVID-19/master/csse_covid_19_data/csse_covid_19_time_series/time_series_covid19_confirmed_global.csv'
df = pd.read_csv(url, index_col=0) # Se importan los datos a la variable df
```

```python
# Previsualización del conjunto de datos en las 10 primeras filas con el método "head()" de pandas
df.head(10)
# En este punto se puede ver la información del conjunto de datos
```

```python
# Con shape, podemos ver las dimensiones del conjunto de datos
# Nota: es probable que la cantidad varíe según el tiempo de ejecución, debido a que esta información es constantemente actualizada.
df.shape
#(filas, columnas)
```

```python
#con info, podemos ver más información del conjunto de datos
print(df.info())
```

```python
# Para el ejercicio, se trabajará con los datos de 'Italia'
# haciendo uso de pandas se seleccionan solo los datos en la columna 'Country/Region' que
# corresponden a 'Italy' y se crea un nuevo dataframe con el nombre "italia"
italia = df.loc[df['Country/Region']=='Italy']
italia.head()
# Se puede ver que cada columna es la cifra de un día particular
```

```python
# Ahora vamos a tomar los datos de Italia pero solo para el año 2022
# Primero, veamos cuantos casos positivos reportados hubo en año 2022
#Ya que cada columna representa la actualización de los casos, es necesario hacer la diferencia del último caso de 2022 menos el primero del mismo año.
column_value_min2022 = italia['1/1/22'].values[0]
column_value_max2022 = italia['12/31/22'].values[0]
print('Primer dato reportado en 2022: ',column_value_min2022)
print('Último dato reportado en 2022: ',column_value_max2022)

#Ahora, obtenemos la cantidad de casos de 2022
globalData2022= column_value_max2022 - column_value_min2022
print('Total de casos reportados en 2022 en Italia: ',globalData2022)
```

```python
#Con el fin de comparar los datos en relación al año 2021 hacemos el mismo ejercicio con el año 2021
column_value_min2021 = italia['1/1/21'].values[0]
column_value_max2021 = italia['12/31/21'].values[0]
print('Primer dato reportado en 2022: ',column_value_min2021)
print('Último dato reportado en 2022: ',column_value_max2021)

#Ahora, obtenemos la cantidad de casos de 2022
globalData2021= column_value_max2021 - column_value_min2021
print('Total de casos reportados en 2021 en Italia: ',globalData2021)
```

```python
# Vamos a utilizar la visualización de Matplotlib para hacer un gráfico de barras y comparar la cantidad de casos reportados
# en ambos años

# Crear una lista con los totales de casos para ambos años
totales = [globalData2021, globalData2022]
# Crear una lista con las etiquetas correspondientes a ambos años
etiquetas = ['2021', '2022']

# Crear la gráfica de barras
fig, ax = plt.subplots()
ax.bar(etiquetas, totales, color=['green', 'red'])# permite configurar el gráfico
ax.set_ylabel('Número de casos')
ax.set_title('Casos de COVID-19 en Italia: 2021 vs 2022')
plt.show()# Se usa para visualizar el gráfico
```

```python
# Con la información que se tiene, se puede definir el promedio de casos diario reportado para 2022

#promedio diario de casos en 2022
media2022Dia=globalData2022/365
print('El promedio de datos positivos por día en 2022: ',media2022Dia.round(2))
# el método round de Python, permite redondear según la cantidad de cifras decimales que se necesiten. En este caso 2
```

```python
# Queremos conocer como fue el comportamiento de los casos reportados como positivos de COVID-19 en el año 2022
# Para esto, primero debemos saber el índice del rango de columnas que queremos estudiar

indiceColumnaMin = italia.columns.get_loc("1/1/22")# índice para el primero de enero de 2022
print("Índice inferior: ",indiceColumnaMin)

indiceColumnaMax = italia.columns.get_loc("12/31/22")# índice para el último día del año 2022
print('Índice Superior: ',indiceColumnaMax)

# Ya que tenemos los índices superior e inferior, podemos hacer un ciclo iterativo que tome los
# datos de todo el año y los almacene un una lista de Python
casosDia2022 = [] #Lista
for i in range(indiceColumnaMin,indiceColumnaMax+1): #rango de iteraciones
    casosDia2022.append(italia.iloc[0,i]-italia.iloc[0,i-1])# en cada iteración se agrega el dato acumulado menos el del día anterior para
    # obtener solo los reportados del día.
# Al final de las iteraciones se obtiene una lista con todos los datos del año día a día
```

```python
# Para visualizar, se utilizará un gráfico de líneas de Matplotlib que reporte los 365 días del año 2022
plt.plot(casosDia2022,
         color='red',
         label='Año 2022'
        ) # Se configura el gráfico; plot se usa para el gráfico de líneas
plt.title('Evolución de casos positivos en el año 2022') # Título
plt.ylabel('Casos confirmados')
plt.xlabel('Día de 2022')
plt.legend()
plt.show()
```

```python
# Vamos a buscar el día del año que reportó mayores casos positivos en el año 2022
diaMayor=max(casosDia2022) # Con la información recogida cada día en la lista, y con el método max se obtiene el mayor dato
print(diaMayor)# Pero solo tenemos el dato, ahora hay que asociarlo a una columna y para eso usamos el índice.

indiceDia=casosDia2022.index(diaMayor)# Con index se obtiene el índice del dato
print('indice del dato mayor: ',indiceDia)

print('Valor del campo con mayor número de reportes: ',italia.iloc[0,indiceDia+indiceColumnaMin])
# Con iloc podemos ver el valor de la fila 0 y la columna del dato buscado.

print('Día de mayor número de datos reportados: '+(df.columns[indiceDia+indiceColumnaMin]))
# Por último, con el método columns() podemos ver la columna a la cual pertenece el dato
# y por consiguiente la fecha del día de más datos reportados en el año 2022
```

```python
# 4. Casos en febrero de 2023
fechas_feb = [col for col in df.columns if '2/1/23' <= col <= '2/28/23']
valor_inicio_feb = españa['2/1/23'].values[0]
valor_final_feb = españa['2/28/23'].values[0]
casos_febrero = valor_final_feb - valor_inicio_feb
print("Casos en febrero:", casos_febrero)
```

```python
# 5. Casos en enero de 2023
fechas_jun = [col for col in df.columns if '1/1/23' <= col <= '1/30/23']
valor_inicio_en = españa['1/1/23'].values[0]
valor_final_en = españa['1/30/23'].values[0]
casos_enero = valor_final_en - valor_inicio_en
print("Casos en enerp:", casos_enero)

```

**Junio no existe en la Base de Datos, por tanto se toma el mes de enero, para poder hacer la gráfica comparativa**

```python
# 6. Gráfico comparativo de febrero vs junio de 2023
plt.bar([ 'Enero 2023','Febrero 2023'], [casos_enero, casos_febrero], color=['skyblue', 'salmon'])
plt.title('Comparación de casos: Febrero vs Junio 2023 (España)')
plt.ylabel('Casos confirmados')
plt.show()
```

**Se evidencia una caida de un cuarto aproximadamente en la cantidad de casos entre enero y febrero en españa de 2023.**

```python
# 7. Promedio diario en marzo de 2023 (del 1 al 9)
# Obtener los casos acumulados para marzo (1 al 9)
fechas_marzo = ['3/1/23', '3/2/23', '3/3/23', '3/4/23', '3/5/23', '3/6/23', '3/7/23', '3/8/23', '3/9/23']
marzo_series = españa[fechas_marzo].values[0]

# Calcular casos diarios restando día a día
marzo_diario = [marzo_series[i+1] - marzo_series[i] for i in range(len(marzo_series)-1)]

# Calcular total y promedio de casos diarios
total_casos_marzo = sum(marzo_diario)
promedio_casos_marzo = total_casos_marzo / len(marzo_diario)

print("Casos diarios del 1 al 9 de marzo:", marzo_diario)
print("Total de casos nuevos del 1 al 9 de marzo:", total_casos_marzo)
print("Promedio diario de nuevos casos:", promedio_casos_marzo)

```

**Durante los primeros nueve días de marzo de 2023 se registraron 7093 casos nuevos en España. Esta cifra se obtiene restando el número de casos acumulados al 1 de marzo del total reportado al 9 de marzo.**

**El promedio diario de nuevos casos en este periodo fue de aproximadamente 886.63 casos por día.**
