```python
import numpy as np

def make_blobs(n_samples, n_features, centers=3, cluster_std=1.0, random_state=42):
    """
    Genera un conjunto de datos aleatorios utilizando la función make_blobs

    Returns:
        X: matriz de datos
        y: vector de clases
    """
    from sklearn.datasets import make_blobs
    return make_blobs(n_samples=n_samples, n_features=n_features, centers=centers, cluster_std=cluster_std, random_state=random_state)

def scatterplot(X, y):
    """
    X contiene los datos y y contiene las etiquetas de clase.
    """
    import matplotlib.pyplot as plt
    plt.figure(figsize=(8, 6))
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.title('Grafico dispersion')
    plt.show()

# Generate some data
X, y = make_blobs(n_samples=150, n_features=2, centers=3, cluster_std=0.5)

# Create a scatterplot
scatterplot(X, y)

print(X)
print(y)
```

```python

import numpy as np

def f(x):
  return np.sin(x)

def df(x):
  return np.cos(x)

def d2f(x):
  return -np.sin(x)

def find_local_min(f, df, d2f, a, b):

  local_min = []

  critical_points = [x for x in np.arange(a, b, 0.1) if df(x) == 0]


  for x in critical_points:
    if d2f(x) > 0:
      local_min.append((x, f(x)))

  return local_min

def find_local_max(f, df, d2f, a, b):

  local_max = []

  critical_points = [x for x in np.arange(a, b, 0.1) if df(x) == 0]


  for x in critical_points:
    if d2f(x) < 0:
      local_max.append((x, f(x)))

  return local_max

a= input("Ingrese inicio de funcion ")
b= input("Ingrese final de funcion ")

local_min = find_local_min(f, df, d2f, a, b)
local_max = find_local_max(f, df, d2f, a, b)

print(local_min)
print(local_max)
```
