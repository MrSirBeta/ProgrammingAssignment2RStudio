```python
!pip install unidecode
import unidecode
import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd

# 1. Cargar shapefile
url = "https://datosabiertos.bogota.gov.co/dataset/856cb657-8ca3-4ee8-857f-37211173b1f8/resource/497b8756-0927-4aee-8da9-ca4e32ca3a8a/download/loca.json"
mapa_bogota = gpd.read_file(url)

# 2. Filtrar y normalizar nombres en df_bogota
df_localidades = df_bogota[df_bogota["Localidad del Hecho"] != "Sin información"].copy()
df_localidades["Localidad_clean"] = df_localidades["Localidad del Hecho"].str.strip().str.upper()
mapa_bogota["LocNombre_clean"] = mapa_bogota["LocNombre"].str.strip().str.upper()

# 3. Agrupar casos por localidad
tabla_localidades = (
    df_localidades.groupby("Localidad_clean")
    .size()
    .reset_index(name="Casos")
)

# 4. Merge con shapefile
mapa_bogota = mapa_bogota.merge(
    tabla_localidades,
    left_on="LocNombre_clean",
    right_on="Localidad_clean",
    how="left"
)
mapa_bogota["Casos"] = mapa_bogota["Casos"].fillna(0)

# 5. Calcular percentil 50%
percentil_50 = mapa_bogota["Casos"].quantile(0.6)

# 6. Graficar mapa completo
fig, ax = plt.subplots(1, 1, figsize=(12,12))
mapa_bogota.plot(
    column="Casos",
    cmap="OrRd",
    linewidth=0.8,
    ax=ax,
    edgecolor="0.8",
    legend=True
)

# 7. Anotar solo localidades por encima del 50%
for idx, row in mapa_bogota.iterrows():
    if row["Casos"] > percentil_50:
        x, y = row['geometry'].centroid.x, row['geometry'].centroid.y
        ax.annotate(
            row['LocNombre'],
            xy=(x, y),
            xytext=(x+0.02, y+0.02),  # ajustar posición del texto
            fontsize=10,
            arrowprops=dict(facecolor='black', arrowstyle='->', lw=1),
            horizontalalignment='left'
        )

ax.set_title("Distribución de casos por localidad - Bogotá", fontsize=15)
ax.axis("off")
plt.show()
```

```python
# 7. Tendencia temporal de todos los casos en Bogotá por mes

# Crear columna Fecha con año y mes (usando formato explícito y loc para evitar warnings)
df_bogota_temporal = df_bogota.copy()

# Map Spanish month names to numbers
month_map = {
    'enero': '01', 'febrero': '02', 'marzo': '03', 'abril': '04',
    'mayo': '05', 'junio': '06', 'julio': '07', 'agosto': '08',
    'septiembre': '09', 'octubre': '10', 'noviembre': '11', 'diciembre': '12'
}
df_bogota_temporal['Mes_Numerico'] = df_bogota_temporal['Mes del hecho'].map(month_map)


df_bogota_temporal["Fecha"] = pd.to_datetime(
    df_bogota_temporal["Año del hecho"].astype(str) + "-" + df_bogota_temporal["Mes_Numerico"] + "-01",
    format="%Y-%m-%d",
    errors="coerce"
)

# Agrupar por mes
tendencia_total = df_bogota_temporal.groupby("Fecha").size().reset_index(name="Casos")

print("Tendencia temporal de todos los casos en Bogotá por mes:")
print(tendencia_total.head())

# Graficar solo si hay datos
if not tendencia_total.empty:
    plt.figure(figsize=(10,6))
    sns.lineplot(x="Fecha", y="Casos", data=tendencia_total, marker="o")
    plt.title("Tendencia mensual de todos los casos en Bogotá")
    plt.xlabel("Fecha")
    plt.ylabel("Número de casos")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
else:
    print("⚠️ No hay datos válidos para graficar la tendencia temporal.")
```

```python
# Agrupar por 'Grupo de edad de la víctima' y 'Sexo de la víctima' y contar los casos
gender_age_counts = df_bogota.groupby(['Grupo de edad de la víctima', 'Sexo de la víctima']).size().reset_index(name='Casos')

print("Distribución de casos por Grupo de edad y Sexo en Bogotá:")
print(gender_age_counts)

# Pivot the table for better visualization
gender_age_pivot = gender_age_counts.pivot(index='Grupo de edad de la víctima', columns='Sexo de la víctima', values='Casos').fillna(0)

# Plotting the distribution of cases by age group and gender
plt.figure(figsize=(12, 7))
gender_age_pivot.plot(kind='bar', stacked=True, figsize=(12, 7))
plt.title("Distribución de casos por Grupo de edad y Sexo en Bogotá")
plt.xlabel("Grupo de edad de la víctima")
plt.ylabel("Número de casos")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Although correlation between categorical variables like gender and age group isn't directly calculated with standard methods like Pearson correlation,
# we can explore the relationship visually or through proportions.
# For example, we can look at the proportion of each gender within each age group.

gender_age_proportions = gender_age_pivot.apply(lambda x: x / x.sum(), axis=1)

print("\nProporción de casos por Grupo de edad y Sexo en Bogotá:")
print(gender_age_proportions)

# Plotting the proportions
plt.figure(figsize=(12, 7))
gender_age_proportions.plot(kind='bar', stacked=True, figsize=(12, 7))
plt.title("Proporción de casos por Grupo de edad y Sexo en Bogotá")
plt.xlabel("Grupo de edad de la víctima")
plt.ylabel("Proporción de casos")
plt.xticks(rotation=45, ha='right')
plt.legend(title="Sexo de la víctima")
plt.tight_layout()
plt.show()
```
